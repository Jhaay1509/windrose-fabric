```python

import requests
import json
from pathlib import Path
from datetime import datetime, timezone

# 1. API endpoint and parameters
BASE_URL = "https://api.open-meteo.com/v1/forecast"

params = {
    "latitude": 55.9533,
    "longitude": -3.1883,
    "hourly": "temperature_2m,relative_humidity_2m",
    "forecast_days": 2,
    "timezone": "Europe/London"
}

# 2. Extract data
response = requests.get(
    BASE_URL,
    params=params,
    timeout=30
)

response.raise_for_status()
weather_data = response.json()

# 3. Basic response validation
if "hourly" not in weather_data:
    raise ValueError("API response has no hourly data")

# 4. Build Bronze storage path
ingestion_date = datetime.now(timezone.utc).date().isoformat()

bronze_dir = Path(
    "/lakehouse/default/Files/bronze/"
    "open_meteo/forecast/edinburgh"
) / f"ingestion_date={ingestion_date}"

bronze_dir.mkdir(parents=True, exist_ok=True)

# 5. Save the raw JSON response
output_path = bronze_dir / "response.json"
output_path.write_bytes(response.content)

# 6. Verify ingestion
print("Extraction successful")
print("City: Edinburgh")
print("Hourly records:", len(weather_data["hourly"]["time"]))
print("Stored at:", output_path)
print("File size (bytes):", output_path.stat().st_size)
```
